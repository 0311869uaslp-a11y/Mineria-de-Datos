<h2>II. Integración: Formato de los datos</h2>

El formateo de datos es el proceso de transformar datos en un formato común, lo que ayuda a los usuarios a realizar comparaciones. Un ejemplo de datos no formateados es el siguiente: se hace referencia a la misma entidad en la misma columna con diferentes valores, como Nueva York y NY.

<h3>Importar los Datos</h3>
En este ejemplo usaremos un historial de tweets que puede obtener en el siguiente enlace:<br/>
    <a href="https://www.trackmyhashtag.com/twitter-dataset">twitter-dataset</a>

```python
import pandas as pd
df = pd.read_csv('tweets.csv')
df.head(5)
```

En este tutorial, soltamos todos los valores faltantes a través de la función <span style="color:green">dropna()</span>

```python
df.dropna(inplace=True)
```

<h3>Tipos de datos incorrectos</h3>
Verificamos que cada columna este asignada al tipo de dato correcto con dtypes

```python
df.dtypes
```

En nuestro caso, podemos convertir la columna <span style="color:blue">Tweet Location</span> en una cadena usando la función <span style="color:green">astype()</span> :

```python
df['Tweet Location'] = df['Tweet Location'].astype('string')
```

Podemos convertir todas las columnas objeto a cadenas con la siguiente instrucción:

```python
import numpy as np
obj_columns = df.select_dtypes(include=object).columns.tolist()
df[obj_columns] = df[obj_columns].astype('string')
```

La función astype() admite todos los tipos de datos descritos en el siguiente enlace:<br/>

<h3>Homogeneizar los datos</h3>
Este aspecto involucra datos categóricos y numéricos. Los datos categóricos deben tener todos el mismo estilo de formato (minúsculas/mayúsculas). Para formatear todos los datos categóricos en minúsculas, podemos usar la siguiente declaración:

```python
df['Tweet Content'] = df['Tweet Content'].str.lower()
```

Otras técnicas para homogeneizar datos categóricos implican los siguientes aspectos:
<ul>
<li>elimine los espacios en blanco en todas partes:<br/>
    df['Tweet Content'] = df['Tweet Content'].str.<span style="color:green">replace</span>(' ', '')</li>
<li>elimine los espacios en blanco al principio de la cadena:<br/>
    df['Tweet Content'] = df['Tweet Content'].str.<span style="color:green">lstrip</span>()</li>
<li>elimine los espacios en blanco al final de la cadena:<br/>
    df['Tweet Content'] = df['Tweet Content']].str.<span style="color:green">rstrip</span>()</li>
<li>elimine los espacios en blanco en ambos extremos: <br/>
    df['Tweet Content'] = df['Tweet Content'].str.<span style="color:green">strip</span>()</li>
</ul>

```python
#a="  hola  mundo  "
#a=a.strip()
#a
df['Tweet Content'] = df['Tweet Content'].str.replace(' ', '')
df['Tweet Content']
```

Los datos numéricos deben tener la misma precisión. Por ejemplo, si queremos 2 puntos decimales, podemos ejecutar el siguiente comando:<br/>
df['User Following'] = df['User Following'].<span style="color:green">round</span>(2)

Otras técnicas para hacer datos numéricos homogéneos incluyen:
<ul>
    <li>Redondear hacia arriba: <br/>
        df['User Following'] = df['User Following'].apply(<span style="color:green">np.ceil</span>)</li>
    <li>Redondear hacia abajo: <br/>
        df['User Following'] = df['User Following'].apply(<span style="color:green">np.floor</span>)</li>
</ul>

<h3>Diferentes valores para un mismo concepto</h3>
Puede ocurrir que un mismo concepto se represente de diferentes formas. Por ejemplo, en nuestro conjunto de datos, la columna Ubicación de Twitter contiene los valores "Columbus,OH" y "Columbus, OH" para describir el mismo concepto. Podemos usar la función <span style="color:green">unique</span>() para enumerar todos los valores de una columna.

```python
df['Tweet Location'].unique()
```

```python
df['Tweet Location'] = df['Tweet Location'].str.replace(' ', '')
df['Tweet Location'].unique()
```

En este caso, solo se redujo 1 valor con replace, para remplazos más especificos y avanzados existen las <span style="color:red">Expresiones Regulares</span>: <br/>
En la siguiente función para cada uno de los estados en la lista se busca ese <span style="color:red">patrón</span> en el texto recibido y si coincide con alguno se regresa ese estado.

```python
import re
def set_pattern(x):
    edos=["Bristol","Brussels","Manchester"]
    for i in edos:
        pattern = r'.*'+i+'.*'
        res = re.match(pattern, x)
        if res:
            return i
    return x

a=" BristoolkjhkjhBrusels"
a=set_pattern(a)
a
```

Ahora podemos aplicar la función a cada valor en la columna <span style="color:blue">'Tweet Location'</span>. Esto se puede lograr usando la función <span style="color:green">apply</span>() combinada con el operador <span style="color:red">lambda</span>. Podemos especificar que la función <span style="color:green">apply</span>() se debe aplicar a cada fila (a través del parámetro eje = 1) y luego a través del operador lambda podemos seleccionar la fila específica y aplicarle la función set_pattern().

```python
df['Tweet Location'] = df.apply(lambda x: set_pattern(x['Tweet Location']), axis=1)
```

```python
df['Tweet Location'].unique()
```
