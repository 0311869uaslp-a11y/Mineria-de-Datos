<h1>I. Preprocesamiento de Datos</h1>
<h2>Normalización</h2>
<p>La normalización es una técnica que se aplica a menudo como parte de la preparación de datos para el aprendizaje automático, ya que permite dar a todas las características el mismo peso. El objetivo de la normalización es cambiar los valores de las <b>columnas numéricas</b> del conjunto de datos para usar una <span style="color:red">escala común</span>, sin distorsionar las diferencias en los intervalos de valores ni perder información. La normalización también es necesaria para que algunos algoritmos modelen los datos correctamente.</p>
<p>Por ejemplo, suponga que el conjunto de datos de entrada contiene una columna con valores comprendidos entre 0 y 1, y otra columna con valores comprendidos entre 10 000 y 100 000. La enorme diferencia en el escala de los números podría generar problemas al intentar combinar los valores como características durante el modelado.</p>
<p>La normalización evita estos problemas mediante la creación de nuevos valores que mantienen la distribución general y las relaciones en los datos de origen, a la vez que se conservan los valores dentro de una escala que se aplica en todas las columnas numéricas que se usan en el modelo.</p>
Algunos métodos de normalización son:
<ul>
    <li>escalado simple</li>
    <li>min max</li>
    <li>z-score</li>
    <li>escala logaritmica</li>
    <li>clipping (Recorte de valores)</li>
</ul>
<p>En este ejemplo, aplicamos cada método a una sola columna. Sin embargo, si desea utilizar cada columna del conjunto de datos como características de entrada de un algoritmo de aprendizaje automático, debe aplicar el <span style="color:purple"><b>mismo método de normalización a todas las columnas</b></span>.</p>
<h3>Importar datos</h3>
Como conjunto de datos de ejemplo, en este tutorial consideramos el conjunto de datos proporcionado por la Protección civil Italiana, relacionado con la cantidad de casos de COVID-19 registrados desde el comienzo de la pandemia de COVID-19. El conjunto de datos se actualiza diariamente y se puede descargar desde <a href="https://raw.githubusercontent.com/pcm-dpc/COVID-19/master/dati-regioni/dpc-covid19-ita-regioni.csv">este enlace</a></p>
<p>Importamos y eliminamos todas las columnas con valores NaN. Esto se hace a través de la función dropna().</p>

```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/pcm-dpc/COVID-19/master/dati-regioni/dpc-covid19-ita-regioni.csv')
df.dropna(axis=1,inplace=True)
df.head(10)
```

Graficamos los datos para visualizar su distribución y escala

```python
import matplotlib.pyplot as plt
plt.hist(df['tamponi'])
```

<h3>Escalado simple</h3>
Convierte cada valor de una columna en un número entre 0 y 1. El nuevo valor se calcula como el valor actual dividido por el valor máximo de la columna.

```python
df['tamponi_single'] = df['tamponi']/df['tamponi'].max()
df['tamponi_single']
```

```python
plt.hist(df['tamponi_single'])
```

<h3>Min Max</h3>
Min Max convierte cada valor de una columna en un número entre 0 y 1. El nuevo valor se calcula como la diferencia entre el valor actual y el valor mínimo, dividido por el rango de los valores de la columna
<img src="img/minmax.png" style="margin:0 auto;"/>

```python
df['totale_casiminmax'] = (df['totale_casi'] - df['totale_casi'].min())/(df['totale_casi'].max() - df['totale_casi'].min())
df['totale_casiminmax']
```

```python
plt.hist(df['totale_casi'])
```

```python
plt.hist(df['totale_casiminmax'])
```

<h3>Escala Logaritmica</h3>
Implica la conversión de una columna a escala logarítmica. Si queremos usar el logaritmo natural, podemos usar la función log() de la biblioteca numpy. Debemos tratar con log(0) porque no existe. Usamos el operador lambda para seleccionar las filas individuales de la columna.

```python
import numpy as np
df['dimessi_guaritilog'] = df['dimessi_guariti'].apply(lambda x: np.log(x) if x != 0 else 0)
```

```python
plt.hist(df['dimessi_guariti'])
```

```python
plt.hist(df['dimessi_guaritilog'])
```

<h3>Clipping</h3>
El recorte de valores (clipping) implica la limitación de todos los valores por debajo o por encima de un determinado valor. El recorte es útil cuando una columna contiene algunos <span style="color:red">valores atípicos</span>. Podemos establecer un valor máximo de vmax y un valor mínimo de vmin y establecer todos los valores atípicos superiores al valor máximo en vmax y todos los valores atípicos inferiores al valor mínimo en vmin. Por ejemplo, podemos considerar la columna ricoverati_con_sintomi y podemos establecer vmax = 10000 y vmin = 10.

```python
vmax = 10000
vmin = 10
df['ricoverati_con_sintomiclip'] = df['ricoverati_con_sintomi'].apply(lambda x: vmax if x > vmax else vmin if x < vmin else x)
df['ricoverati_con_sintomiclip']
```

```python
plt.hist(df['ricoverati_con_sintomi'])
```

```python
plt.hist(df['ricoverati_con_sintomiclip'])
```

<h2>Estandarización: Puntajes Z</h2>
<p>La estandarización a menudo se confunde con la normalización, sin embargo, se refieren a cosas diferentes. La normalización implica ajustar los valores medidos en diferentes escalas a una escala común, mientras que la estandarización transforma los datos para que tengan una <span style="color:red">media</span> de cero y una desviación estándar de 1. La estandarización también se realiza a través de una transformación z-score, donde el nuevo valor se calcula como la diferencia entre el valor actual y el valor promedio, dividido por la <span style="color:red">desviación estándar</span>.</p>
<p>Z-score es una medida estadística que especifica qué tan lejos está un único punto del resto del conjunto de datos. Como destaca Mahbubul Alam en su artículo, el z-score se puede usar para detectar <span style="color:red">valores atípicos</span> en un conjunto de datos.</p> 
<img src="img/zscore.png" style="margin:0 auto;"/>
<ul>
    <li>El valor promedio de una columna se puede obtener a través de la función mean()</li>
    <li>la desviación estándar a través de la función std()</li>
</ul>
La puntuación z te indica cuántas <span style="color:red">desviaciones estándar</span> x está el valor por encima o por debajo la <span style="color:red">media</span>, &mu;.
<img src="img/empirical_rule.jpg" width="480px" style="margin:0 auto;"/>
Los valores típicos obtenidos por una transformación de z-score oscilan entre -3 y 3.

```python
df['decedutizscore'] = (df['deceduti']-df['deceduti'].mean())/df['deceduti'].std()
df['decedutizscore']
```

Alternativamente también se puede obtener usando la libreria scipy asi:

```python
from scipy.stats import zscore
df['zscore-deceduti'] = zscore(df['deceduti'])
df['zscore-deceduti']
```

```python
plt.hist(df['deceduti'])
```

```python
plt.hist(df['decedutizscore'])
```

```python
print("min: ",df['decedutizscore'].min(),"\nmax: ",df['decedutizscore'].max())
```

<h3>Detectar valores atípicos</h3>
La estandarización se puede utilizar para detectar y eliminar valores atípicos. Por ejemplo, se puede definir un umbral para especificar qué valores pueden considerarse atípicos. En este ejemplo, establecemos el umbral = 2. Podemos agregar una nueva columna al dataframe, llamada valores atípicos, que se establece en Verdadero si el valor es menor que -2 o mayor que 2. Usamos la función numpy where() para realizar comparaciones

```python
import numpy as np
threshold = 2
df['outliers'] = np.where((df['decedutizscore'] - threshold > 0), True, np.where(df['decedutizscore'] + threshold < 0, True, False))
```

Removemos los valores atípicos usando drop

```python
df.drop(df[df['outliers'] == True].index,inplace=True)
plt.hist(df['decedutizscore'])
```

<h3>Resumen</h3>
En este ejemplo, revisamos las diferentes técnicas para realizar la normalización de datos: escalado simple, min max, z-score, escala logaritmica y recorte de valores. Entonces, la pregunta es: ¿cuál es la mejor técnica? En realidad, no existe una técnica mejor que las demás, la elección de un método u otro depende de lo que queramos como salida. De este modo:
<ul>
    <li>si desea una salida entre 0 y 1, debe usar escalado simple o min max.</li>
    <li>si desea una salida alrededor de 0, que también incluya valores negativos, debe usar z-score</li>
    <li>si sus datos contienen muchos valores atípicos, puede usar el recorte</li>
    <li>si desea cambiar la escala de sus datos, puede usar la escala logaritmica.</li>
</ul>
La Función logística es muy usada por <span style="color:red">redes neuronales</span> como función de activación.<br/>
También revisamos la diferencia entre normalización y estandarización. La normalización de alguna manera incluye la estandarización.
La estandarización se puede utilizar para detectar y eliminar valores <span style="color:red">atípicos</span> de un conjunto de datos.
