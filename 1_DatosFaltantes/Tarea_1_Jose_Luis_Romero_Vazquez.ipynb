8. Remplace los valores faltantes de las columnas numéricas por la media aritmética

```python
import numpy as np
numeric = df.select_dtypes(include=np.number)
numeric_columns = numeric.columns
```

```python
df[numeric_columns] = df[numeric_columns].fillna(df.mean(numeric_only= True))
```

10.Remplace los valores faltantes de la columna categórica "STDs (number)" por la 
moda de esa columna

```python
boolean_columns = df.select_dtypes(include=object).columns.tolist()
boolean_columns.remove('STDs (number)')
```

```python
#df["steroid"]=df["steroid"].fillna(df["steroid"].mode()[0])
for i in boolean_columns:
    df[i]=df[i].fillna(df[i].mode()[0])
```

12.Vuelva a cargar los datos y realice una interpolación para remplazar los datos 
faltantes de las columnas numéricas

```python
import pandas as pd
df = pd.read_csv('cervical-cancer_csv.csv')
df.head(10)
```

```python
df.isna().sum()/len(df)*100
```

```python
numeric = df.select_dtypes(include=np.number)
numeric_columns = numeric.columns
df.head(10)
```

```python
df[numeric_columns] = df[numeric_columns].interpolate(method ='linear', limit_direction ='forward')
```
