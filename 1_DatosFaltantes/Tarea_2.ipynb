```python
Q1=df["albumin"].quantile(0.25)
print("Primer cuartil:\t\t",Q1)
mediana=df["albumin"].median()
print("Mediana:\t\t",mediana)
Q3=df["albumin"].quantile(0.75)
print("Tercer cuartil:\t\t",Q3)
IQR=Q3-Q1
print("Rango intercuartil:\t",IQR)
minimo=df["albumin"].min()
print("Valor mínimo:\t\t",minimo)
maximo=df["albumin"].max()
print("Valor máximo:\t\t",maximo)
```

5. Obtenga los valores de los bigotes inferior y superior.

```python
BI=Q1-1.5*IQR
BS=Q3+1.5*IQR
print("Bigote inferior:\t",BI)
print("Bigote superior:\t",BS)
```

```python
ubicacion_outliers=(df["albumin"]<BI) | (df["albumin"]>BS)
print("Ubicacion outliers:\n",ubicacion_outliers)
```

7. Consulte el valor de los outliers.

```python
outliers=df[ubicacion_outliers]
outliers
```
