<h1>Preprocesamiento de datos</h1>
Este tutorial explica cómo preprocesar datos usando la biblioteca pandas. El preprocesamiento es el proceso de hacer un análisis previo de los datos, con el fin de transformarlos en un formato estándar y normalizado.
Realizaremos las siguientes actividades del preprocesamiento:
<ol>
<li>valores faltantes</li>
<li>estandarización de datos</li>
<li>normalización de datos</li>
<li>agrupación de datos</li>
<a href="https://datahub.io/machine-learning/hepatitis">Repositorio datos</a>

<h2>I. Valores Faltantes</h2>

<h3>Importar Datos</h3>
Lo primero es incluir la librería Pandas y convertir los datos en un dataframe para su fácil manipulación. A través del método head(10) imprimimos solo las primeras 10 filas del conjunto de datos

```python
import pandas as pd
df = pd.read_csv('hepatitis.csv')
df.head(10)
```

<h3>Identificar valores faltantes</h3>
Observamos que el conjunto de datos presenta algunos problemas. Por ejemplo, la columna <span style="color:blue">alk_phosphate</span> no está disponible para todas las filas. En algunos casos presenta el valor <span style="color:red">NaN</span>, lo que significa que falta el valor.

Para verificar si nuestro conjunto de datos contiene valores faltantes, podemos usar la función <span style="color:green">isna()</span>, que devuelve si una celda del conjunto de datos es NaN o no. Entonces podemos contar cuántos valores faltantes hay para cada columna.

```python
df.isna().sum()
```

Ahora podemos contar el porcentaje de valores faltantes para cada columna, simplemente dividiendo el resultado anterior por la longitud del conjunto de datos (<span style="color:green">len(df)</span>) y multiplicando por 100.

```python
df.isna().sum()/len(df)*100
```

Cuando se trata de valores perdidos, se pueden aplicar diferentes alternativas:
<ul>
<li>comprobar la fuente, por ejemplo, poniéndose en contacto con la fuente de datos para corregir los valores que faltan</li>
<li>descartar valores faltantes</li>
<li>reemplazar el valor que falta con un valor</li>
<li>deje el valor que falta como está.</li>
</ul>

<h3>Eliminar valores faltantes</h3>
Para descartar los valores faltantes tiene las siguientes alternativas:
<ul>
<li>eliminar filas que tienen valores faltantes</li>
<li>elimine toda la columna que contiene valores faltantes.<br/>
    Podemos usar <span style="color:green">dropna()</span> especificando el eje a considerar.
    <ul>
        <li>Si configuramos <span style="color:green">axis = 0</span>, descartamos toda la fila,</li>
        <li>si configuramos <span style="color:green">axis = 1</span>, descartamos toda la columna. </li>
        <li>Si aplicamos la función <span style="color:green">df.dropna(axis=0)</span> quedan 80 filas del conjunto de datos.</li>
        <li>Si aplicamos la función <span style="color:green">df.dropna(axis=1)</span>, solo quedan las columnas edad, sexo, antivirales, histología y clase. </li>
    </ul>
    Sin embargo, los valores eliminados no se aplican al marco de datos original, sino solo al resultado. <br/>
    Podemos usar el argumento <span style="color:green">inplace=True</span> para almacenar cambios en el marco de datos original df (<span style="color:green">df.dropna(axis=1,inplace=True)</span>).
</ul>

```python
df.dropna(axis=1)
```

Como alternativa, podemos especificar solo la columna en la que se debe aplicar la operación de eliminación. En el siguiente ejemplo, solo se consideran las filas faltantes relacionadas con la columna <span style="color:green">liver_big</span>. Esto se puede lograr a través del parámetro subconjunto, que permite especificar el subconjunto de columnas donde aplicar la operación de eliminación.

```python
df.dropna(subset=['liver_big'],axis=0,inplace=True)
```

Ahora podemos verificar si todavía faltan valores para el resto de las columnas.

```python
df.isna().sum()/len(df)*100
```

Otra alternativa implica la eliminación de columnas donde está disponible un cierto porcentaje de valores no nulos. Esto se puede lograr a través del parámetro de umbral <span style="color:green">thresh</span>. En el siguiente ejemplo, mantenemos solo las columnas donde hay al menos el 80% de los valores no nulos.

```python
df.dropna(thresh=0.8*len(df),axis=1,inplace=True)
```

```python
df
```

<h3>Reemplazar valores faltantes</h3>
Una buena estrategia cuando se trata de valores faltantes implica su sustitución por otro valor. Por lo general, se adoptan las siguientes estrategias:
<ul>
    <li>para valores numéricos, reemplace el valor faltante con el valor promedio de la columna</li>
    <li>para valores categóricos, reemplace el valor faltante con el valor más frecuente de la columna</li>
    <li>utilizar otras funciones</li>
</ul>
Para reemplazar los valores faltantes, se pueden usar tres funciones: fillna(), replace() e interpolate().<br/>
La función <span style="color:green">fillna()</span> reemplaza todos los valores de NaN con el valor pasado como argumento. Por ejemplo, para valores numéricos, todos los valores de NaN en las columnas numéricas podrían reemplazarse con el valor promedio. Para enumerar el tipo de una columna, podemos usar el atributo <span style="color:green">dtypes</span>:

```python
df.dtypes
```

<h4>Columnas numéricas</h4>
Primero seleccionamos las columnas numéricas

```python
import numpy as np
numeric = df.select_dtypes(include=np.number)
numeric_columns = numeric.columns
```

Luego, llenamos los valores NaN de las columnas numéricas con el valor promedio, dado por la función <span style="color:green">df.mean()</span>

```python
df[numeric_columns] = df[numeric_columns].fillna(df.mean(numeric_only= True))
```

Comprobemos que se han eliminado los valores de NaN en las columnas numéricas.

```python
df.isna().sum()/len(df)*100
```

<h4>Columnas categóricas</h4>
Observamos que en dtypes las columnas categóricas se describen como objetos. Así podemos seleccionar las columnas del objeto. Nos gustaría considerar solo columnas booleanas. Sin embargo, el tipo de objeto incluye también la clase de columna, que es una cadena. Seleccionamos todas las columnas de objetos y luego eliminamos de ellas la clase de columna. Entonces podemos convertir el tipo del resultado a bool.<br/>
sex no es bool 

```python
boolean_columns = df.select_dtypes(include=object).columns.tolist()
boolean_columns.remove('class')
boolean_columns.remove('sex')
```

Ahora podemos reemplazar todos los valores que faltan en columnas booleanas con el valor más frecuente. Podemos usar la función <span style="color:red">mode()</span> para calcular el valor más frecuente. Usamos la función fillna() para reemplazar los valores faltantes, pero también podríamos usar la función <span style="color:green">replace(old_value,new_value)</span>.

```python
#df["steroid"]=df["steroid"].fillna(df["steroid"].mode()[0])
for i in boolean_columns:
    df[i]=df[i].fillna(df[i].mode()[0])
```

```python
df.to_csv("hepatitis_sinNAN.csv",index=False)
```

Verificamos que no queda ningun valor Nan

```python
df.isna().sum()/len(df)*100
```

<h3>Interpolación</h3>
Otra solución para reemplazar los valores faltantes implica el uso de otras funciones, como la <span style="color:red">interpolación lineal</span>. En este caso, por ejemplo, podríamos reemplazar un valor faltante sobre una columna, con la interpolación entre el anterior y el siguiente. Esto se puede lograr mediante el uso de la función <span style="color:green">interpolate()</span>.

Como ya hemos gestionado todos los valores que faltan, recargamos el conjunto de datos.

```python
df = pd.read_csv('hepatitis.csv')
df.isna().sum()/len(df)*100
```

Seleccionamos solo los valores numéricos

```python
numeric = df.select_dtypes(include=np.number)
numeric_columns = numeric.columns
df.head(10)
```

Ahora podemos aplicar la función de interpolación () a columnas numéricas, configurando también la dirección límite para avanzar. Esto significa que la interpolación lineal se aplica desde la primera fila hasta la última.

```python
df[numeric_columns] = df[numeric_columns].interpolate(method ='linear', limit_direction ='forward')
```

Por ejemplo, en la línea 6 la columna bilirrubin, que era NaN antes de la interpolación, ahora asume el valor 0.95, que es la interpolación entre 0.90 (línea 5) y 1.00 (línea 7).

```python
df.head(10)
```

<h3>Resumen</h3>
Los datos faltantes pueden alterar el proceso de análisis de datos, por lo que deben gestionarse.<br/>
Se pueden utilizar tres estrategias para tratar los datos faltantes:
<ul>
    <li>descartar datos faltantes: esto se puede hacer cuando el conjunto de datos tiene una pequeña cantidad de datos faltantes</li>
    <li>reemplazar los datos que faltan con otros valores, como la media o el valor más frecuente</li>
    <li>deje los datos faltantes como están.</li>
</ul>
