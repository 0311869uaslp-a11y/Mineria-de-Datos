<h1 style="margin:0px 0px 20px;">Métricas para medir distancia</h1>
<p>Muchos algoritmos, ya sean supervisados o no supervisados, utilizan medidas de distancia y saber cuándo usar alguna medida de distancia especifica puede ayudarlo a pasar de un clasificador deficiente a un modelo preciso.</p>
<p>El día de hoy, repasaremos muchas medidas de distancia y exploraremos cómo y cuándo se pueden usar mejor. Lo que es más importante, hablaremos sobre sus desventajas para que pueda reconocer cuándo evitar ciertas medidas.</p>
<p><b>NOTA:</b> Para la mayoría de las medidas de distancia, se podrían y se han escrito documentos largos y elaborados sobre sus casos de uso, ventajas y desventajas. Cubriremos tanto como sea posible, pero considere estas notas como una visión global de estas medidas.
</p>
<ol>
    <li><b>Distancia Euclidiana</b> $O(n)$<br/>
    <img src="img/euclidiana.png" width="120px" style="display:block;margin:0 auto;"><br/>Longitud de segmento que conecta dos puntos. Su fórmula es el teorema de Pitágoras 
        $$D(x,y)=\sqrt{\sum_{i=1}^{n}(x_i -y_i)^2}$$
        <h5>Desventajas</h5>
Aunque es una medida de distancia común, la distancia euclidiana no es una escala invariante, lo que significa que las distancias calculadas pueden estar sesgadas según las unidades de las entidades. Por lo general, uno necesita <span style="color:red">normalizar</span> los datos antes de usar esta medida de distancia.<br/>
Además, a medida que aumenta la <span style="color:red">dimensionalidad</span> de sus datos, la distancia euclidiana se vuelve menos útil. Esto tiene que ver con la maldición de la dimensionalidad que se relaciona con la noción de que el espacio de dimensiones superiores no actúa como esperaríamos, intuitivamente, del espacio bidimensional o tridimensional. Para un buen resumen, vea <a href="https://stats.stackexchange.com/questions/99171/why-is-euclidean-distance-not-a-good-metric-in-high-dimensions">este post</a>.<br/>
        También es sensible a datos atípicos.
    <h5>Casos de uso</h5>
La distancia euclidiana funciona muy bien cuando tiene datos de baja dimensión y es importante medir la magnitud de los vectores. Métodos como <span style="color:red">kNN y HDBSCAN</span> muestran excelentes resultados listos para usar si se usa la distancia euclidiana en datos de baja dimensión.
<p>Aunque se han desarrollado muchas otras medidas para tener en cuenta las desventajas de la distancia euclidiana, sigue siendo una de las medidas de distancia más utilizadas por buenas razones. Es increíblemente intuitivo de usar, simple de implementar y muestra excelentes resultados en muchos casos de uso, K-means y fuzzy C-means por mencionar algunos.</p>
    </li>
    <div style="height:30px"></div>
    <li><b>Distancia Manhattan</b> $O(n)$<br/>
        <div style="display:block;margin:0 auto;width:320px;">
            <img src="img/manhattan.png" width="120px" style="margin-right:60px"/>
            <img src="img/taxicab.png" width="120px"/>
        </div>
        La distancia de Manhattan, a menudo llamada distancia de Taxi o distancia de bloques de ciudad, calcula la distancia entre vectores de valor real. Imagine vectores que describen objetos en una cuadrícula uniforme, como un tablero de ajedrez. La distancia de Manhattan se refiere a la distancia entre dos vectores si solo pudieran moverse en ángulo recto. No hay movimiento diagonal involucrado en el cálculo de la distancia.
$$D(x,y)=\sum_{i=1}^{k}\lvert x_i - y_i \rvert$$

<h5>Desventajas</h5>
Aunque la distancia de Manhattan parece funcionar bien para datos de alta dimensión, es una medida algo menos intuitiva que la distancia euclidiana, especialmente cuando se usa en datos de alta dimensión.<br/>
Además, es más probable que proporcione un valor de distancia más alto que la distancia euclidiana, ya que no ofrece el camino más corto posible. Esto no necesariamente da problemas, pero es algo que debe tener en cuenta.
<h5>Casos de uso</h5>
Cuando su conjunto de datos tiene atributos discretos y/o binarios, Manhattan parece funcionar bastante bien, ya que tiene en cuenta las rutas que, de manera realista, podrían tomarse dentro de los valores de esos atributos. Tomar la distancia euclidiana, por ejemplo, crearía una línea recta entre dos vectores cuando en realidad esto podría no ser posible.</li>
    <div style="height:30px"></div>
    <li><b>Distancia Chebyshev</b> $O(n)$<br/>
<img src="img/chebyshev.png" width="120px" style="display:block;margin:0 auto;"/><br/>
La distancia de Chebyshev se define como la mayor diferencia entre dos vectores a lo largo de cualquier dimensión de coordenadas. En otras palabras, es simplemente la distancia máxima a lo largo de un eje. Debido a su naturaleza, a menudo se la denomina distancia del tablero de ajedrez, ya que el número mínimo de movimientos que necesita un rey para ir de una casilla a otra es igual a la distancia de Chebyshev.
$$D(x,y)=max_i(\lvert x_i - y_i \rvert)$$

<h5>Desventajas</h5>
Chebyshev generalmente se usa en casos de uso muy específicos. Por esa razón, se sugiere usarlo solo cuando esté absolutamente seguro de que se adapta a sus necesidades.

<h5>Casos de uso</h5>
Como se mencionó anteriormente, la distancia de Chebyshev se puede usar para extraer el número mínimo de movimientos necesarios para ir de un cuadrado a otro. Además, puede ser una medida útil en juegos que permiten un movimiento de 8 direcciones sin restricciones.<br/>
En la práctica, la distancia de Chebyshev se usa a menudo en la logística del almacén, ya que se parece mucho al tiempo que tarda una puente-grúa en mover un objeto.</li>
    <div style="height:30px"></div>
    <li><b>Distancia Minkowski</b><br/>
<img src="img/minkowski.png" width="150px" style="display:block;margin:0 auto;"/><br/>
La distancia de Minkowski es una medida un poco más compleja que la mayoría. Es una métrica utilizada en el espacio vectorial normado (espacio real n-dimensional), lo que significa que se puede utilizar en un espacio donde las distancias se pueden representar como un vector que tiene una longitud.<br/>
Esta medida tiene tres requisitos:
        <ul>
            <li><b>Vector cero</b>: el vector cero tiene una longitud de cero, mientras que todos los demás vectores tienen una longitud positiva. Por ejemplo, si viajamos de un lugar a otro, entonces esa distancia siempre es positiva. Sin embargo, si viajamos de un lugar a sí mismo, entonces esa distancia es cero.</li>
            <li><b>Factor escalar</b>: cuando multiplica el vector con un número positivo, su longitud cambia mientras mantiene su dirección. Por ejemplo, si recorremos cierta distancia en una dirección y sumamos la misma distancia, la dirección no cambia.</li>
            <li><b>Desigualdad triangular</b>:la distancia más corta entre dos puntos es una línea recta.</li>
        </ul>
        $$D(x,y)=\left( \sum_{i=1}^{n}\lvert x_i - y_i \rvert ^p \right) ^\frac{1}{p}$$
Lo más interesante de esta medida de distancia es el uso del parámetro p. Podemos usar este parámetro para manipular las métricas de distancia para que se parezcan mucho a otras.Valores comunes de p:
        <ul>
            <li>p=1 distancia Manhattan</li>
            <li>p=2 distancia Euclidiana</li>
            <li>p=$\infty$ Distancia Chebysev</li>
        </ul>
<h5>Desventajas</h5>
Minkowski tiene las mismas desventajas que las medidas de distancia que representan, por lo que una buena comprensión de métricas como la distancia de Manhattan, euclidiana y Chebyshev es extremadamente importante.<br/>
Además, el parámetro p en realidad puede ser problemático para trabajar, ya que encontrar el valor correcto puede ser bastante ineficiente desde el punto de vista computacional según su caso de uso.

<h5>Casos de uso</h5>
La ventaja de p es la posibilidad de iterar sobre él y encontrar la medida de distancia que funcione mejor para su caso de uso. Le permite una gran flexibilidad sobre su métrica de distancia, lo que puede ser un gran beneficio si está familiarizado con p y muchas medidas de distancia.</li>
    <div style="height:30px"></div>
    <li><b>Distancia Euclidiana Cuadrada SED</b> $O(n)$<br/>
    Calcular distancias usando este enfoque evita la necesidad de usar la función de raíz cuadrada. Como su nombre lo refleja, la SED es igual a la distancia euclidiana al cuadrado. Por lo tanto, SED puede reducir el trabajo computacional al calcular distancias entre observaciones. Por ejemplo, se puede usar en agrupamiento, clasificación, procesamiento de imágenes y otros dominios.
    $$D(x,y)=\sum_{i=1}^{n}(x_i -y_i)^2$$</li>
    <div style="height:30px"></div>
    <li><b>Distancia Canberra</b> $O(n)$<br/>
Es una versión ponderada de la distancia de Manhattan utilizada en agrupamiento, agrupamiento difuso, clasificación, seguridad informática y sistemas de detección de spam. Es más resistente a los valores atípicos en contraste con la métrica Manhattan.
        $$D_{canb}=\sum_{i=1}^{n}\frac{\lvert x_i-y_i \rvert}{(x_i+y_i)}$$
    </li>
    <div style="height:30px"></div>
    <li><b>Distancia Chi-cuadrada</b> $O(n)$<br/>
    La distancia chi-cuadrada se usa comúnmente en la <span style="color:red">visión por computadora</span> al realizar análisis de textura para encontrar (des) similitudes entre histogramas normalizados, lo que se conoce como "coincidencia de histogramas" (<span style="color:green">Histogram matching</span>).<br/>
    Un algoritmo de reconocimiento facial sería un gran ejemplo que utiliza esta métrica para comparar dos histogramas. Por ejemplo, en el paso de predicción de una nueva cara, el modelo calcula el histograma de la imagen recién capturada, lo compara con los histogramas guardados (a menudo almacenados en un archivo .yaml) y luego trata de encontrar la mejor coincidencia. Esta comparación se realiza mediante el cálculo de la distancia Chi-cuadrada entre cada par de histogramas de n contenedores.
    $$D(x,y)=\sum_{i=1}^{n}\frac{(x_i-y_i)^2}{x_i+y_i}$$
    Como sabrá, esta fórmula es diferente de la prueba estadística <span style="color:red">Chi-cuadrada</span> para distribuciones normales estándar, donde se usa para decidir si se mantiene o rechaza la hipótesis nula</li>
    <div style="height:30px"></div>
    <li><b>Distancia Euclidiana Ponderada</b> $O(n)$<br/>
    $$D(x,y)=\sqrt{\sum_{i=1}^{n}(w_i (x_i -y_i))^2}$$ donde $$w_i=\frac{1}{\sigma_i}$$
    siendo $\sigma_i^2$ una estima de la varianza del i-ésimo valor x_i. Aquí, los datos que son menos confiables (con mayor varianza) son pesados menos.<br/>
    La matriz de peso permite aumentar el efecto de los puntos de datos más importantes que el de los menos importantes.
    <h5>Desventajas</h5>
    Las variables contribuyen de forma independiente a la medida de la distancia. Los valores redundantes podrían dominar la similitud entre los puntos de datos
    <h5>Casos de uso</h5>
    Usada en el <span style="color:red">agrupamiento difuso C-means</span>.</li>
</ol>

```python
import pandas as pd # dataframe manipulation
import numpy as np # linear algebra

# data visualization
import matplotlib.pyplot as plt
%matplotlib inline
from yellowbrick.cluster import KElbowVisualizer # cluster visualizer

# sklearn kmeans
from sklearn.cluster import KMeans
from sklearn.metrics.cluster import contingency_matrix

# pyclustering kmeans
from pyclustering.cluster.kmeans import kmeans
from pyclustering.utils.metric import distance_metric
from pyclustering.cluster.center_initializer import random_center_initializer
from pyclustering.cluster.encoder import type_encoding
from pyclustering.cluster.encoder import cluster_encoder
```

<h3>Cargar los datos</h3>

```python
iris = pd.read_csv('Iris.csv')
iris.head()
```

<h3>Separar valores de la clase</h3>

```python
X = iris.iloc[:, 1:5].values # feature matrix
y = iris.iloc[:, -1].values # class matrix
```

<h3>Método Elbow para determinar número de clusters optimo</h3>

```python
# Instantiate the clustering model and visualizer
model = KMeans(n_init=50)
visualizer = KElbowVisualizer(model, k=(1, 11))

visualizer.fit(X) # Fit the data to the visualizer
visualizer.show() # Finalize and render the figure
plt.show()
```

<h3>K-means</h3>

```python
# instatiate KMeans class and set the number of clusters
km_model = KMeans(n_clusters=3, random_state=10, n_init=50)

# call fit method with data 
km = km_model.fit_predict(X)

# coordinates of cluster center
centroids = km_model.cluster_centers_ 

# cluster label for each data point
labels = km_model.labels_ 
```

```python
plt.figure(figsize=(8,6))
plt.scatter(
    X[km == 0, 0], X[km == 0, 1],
    s=25, c='lightgreen',
    marker='s', edgecolor='black',
    label='cluster 1'
)

plt.scatter(
    X[km == 1, 0], X[km == 1, 1],
    s=25, c='yellow',
    marker='o', edgecolor='black',
    label='cluster 2'
)

plt.scatter(
    X[km == 2, 0], X[km == 2, 1],
    s=25, c='lightblue',
    marker='v', edgecolor='black',
    label='cluster 3'
)

# visualise centroids
plt.scatter(
    centroids[:, 0], centroids[:, 1],
    s=300, marker='*',
    c='red', edgecolor='black',
    label='centroids'
)
plt.legend(scatterpoints=1)
plt.grid()
plt.show()
```

<h3>Informar puntuación de pureza</h3>
Aquí, la puntuación de Pureza se calcula utilizando Ground Truth o External accessment. Dado que el conjunto de datos cuenta con una etiqueta real, la precisión de la etiqueta de clúster calculada con KMeans se compara con la clase de etiqueta verdadera.

```python
def purity_score(y_true, y_pred):
    # compute contingency matrix (also called confusion matrix)
    confusion_matrix = contingency_matrix(y_true, y_pred)
    # return purity
    return np.sum(np.amax(confusion_matrix, axis=0)) / np.sum(confusion_matrix)

# Report Purity Score
purity = purity_score(y, labels)
print(f"The purity score is {round(purity*100, 2)}%")
```

<h3>Pyclustering para usar diferentes métricas de distancia en K-means</h3>

Probaremos las métricas de distancia como Manhattan, Chebyshev, euclidiana, etc.
La distancia de Minkowski es solo la generalización de la distancia euclidiana (p=2), manhattan (p=1) y chebyshev (p=Inf).
Aunque para pyclustering, los centros iniciales se pueden inicializar usando el algoritmo k ++, pero he inicializado aleatoriamente el centro inicial para comparar los resultados con sklearn kmeans.
La pureza más alta en este caso se logra utilizando <span style="color:red">Chi-Square</span> como la métrica de distancia.

```python
import math
def distanciaE(point1, point2):
     distance = 0.0
     for i in range(len(point1)):
         distance += (point1[i] - point2[i]) ** 2.0
     return math.sqrt(distance)
def distanciaAVE(point1, point2):
     distance = 0.0
     for i in range(len(point1)):
         distance += (point1[i] - point2[i]) ** 2.0
     return math.sqrt(distance/len(point1))

def distanciaWE(point1, point2):
     #W=(0.4,0.3,0.2,0.1) #86.67
     #W=(0.3,0.2,0.4,0.1) #89.33   
     #W=(0.1,0.2,0.3,0.4) #91.33
     W=(0.1,0.1,0.1,0.7)  #96
     distance = 0.0
     for i in range(len(point1)):
         distance += W[i]*(point1[i] - point2[i]) ** 2.0
     return math.sqrt(distance)

# define dictionary for distance measures
distance_measures = {'Euclidiana': 0, 'Euclidiana cuadrada': 1, 'Manhattan': 2, 'Chebyshev': 3, 
                    'Canberra': 5, 'Chi-square': 6,'propia':1000}

# function defined to compute purity score using pyclustering for various distance measures
def pyPurity(dist_measure):
    initial_centers = random_center_initializer(X, 3, random_state=5).initialize()
    # instance created for respective distance metric
    if dist_measure!=1000:
        instanceKm = kmeans(X, initial_centers=initial_centers, metric=distance_metric(dist_measure))
    else:
        instanceKm = kmeans(X, initial_centers=initial_centers, metric=distance_metric(1000, func=distanciaWE))
    # perform cluster analysis
    instanceKm.process()
    # cluster analysis results - clusters and centers
    pyClusters = instanceKm.get_clusters()
    pyCenters = instanceKm.get_centers()
    # enumerate encoding type to index labeling to get labels
    pyEncoding = instanceKm.get_cluster_encoding()
    pyEncoder = cluster_encoder(pyEncoding, pyClusters, X)
    pyLabels = pyEncoder.set_encoding(0).get_clusters()
    # function purity score is defined in previous section
    return purity_score(y, pyLabels)

# print results
for measure, value in distance_measures.items():
    print(f"Purity score con distancia {measure}: {round(pyPurity(value)*100, 2)}%")
```

    <li><b>Similitud del coseno</b> $O(3n)$<br/>
<img src="img/coseno.png" width="120px" style="display:block;margin:0 auto;"/><br/>
La similitud del coseno se ha utilizado a menudo como una forma de contrarrestar el problema de la distancia euclidiana con alta dimensionalidad. La similitud del coseno es simplemente el coseno del ángulo entre dos vectores. También tiene el mismo producto interno de los vectores si se normalizaran para que ambos tuvieran longitud uno.
<p>Dos vectores con exactamente la misma orientación tienen una similitud de coseno de 1, mientras que dos vectores diametralmente opuestos entre sí tienen una similitud de -1. Tenga en cuenta que su magnitud no tiene importancia ya que es una medida de orientación.</p>
$$D(x,y)=cos(\theta)=\frac{x\cdot y}{\lVert x \rVert \lVert y \rVert}$$
    <h5>Desventajas</h5>
Una de las principales desventajas de la similitud del coseno es que no se tiene en cuenta la magnitud de los vectores, sino simplemente su dirección. En la práctica, esto significa que las diferencias de valores no se tienen plenamente en cuenta. Si toma un sistema de recomendación, por ejemplo, la similitud del coseno no tiene en cuenta la diferencia en la escala de calificación entre diferentes usuarios.
<h5>Casos de uso</h5>
<p>Usamos la similitud del coseno a menudo cuando tenemos datos de alta dimensión y cuando la magnitud de los vectores no es importante. Para los <span style="color:red">análisis de texto</span>, esta medida se utiliza con bastante frecuencia cuando los datos se representan mediante recuentos de palabras. Por ejemplo, cuando una palabra aparece con más frecuencia en un documento que en otro, esto no significa necesariamente que un documento esté más relacionado con esa palabra. Puede darse el caso de que los documentos tengan longitudes desiguales y la magnitud del conteo sea de menor importancia. Entonces, podemos usar mejor la similitud del coseno que no tiene en cuenta la magnitud.</p></li>
    <div style="height:30px"></div>
    <li><b>Distancia de Mahalanobis </b>$O(3n)$<br/>
<img src="img/mahalanobis.png" width="480px" style="display:block;margin:0 auto;"/><br/>
    La distancia de Mahalanobis es una medida de la distancia entre un punto P y una distribución D. La idea de medir es a cuántas desviaciones estándar está P de la media de D.<br/>
El beneficio de usar la distancia mahalanobis es que tiene en cuenta la <span style="color:green">covarianza</span>, lo que ayuda a medir la fuerza/similitud entre dos objetos de datos diferentes. La distancia entre una observación y la <span style="color:green">media</span> se puede calcular de la siguiente manera:
    $$D_M(x)=\sqrt{(x-\mu)^TS^{-1}(x-\mu)}$$
    Aquí, S es la métrica de covarianza. Estamos usando la inversa de la métrica de covarianza para obtener una ecuación de distancia normalizada por varianza.
    <h5>Desventajas</h5>
    Puede ser costoso en términos computacionales $O(3n)$.
    <h5>Casos de uso</h5>
    Es una medida basada en datos que puede aliviar la distorsión de la distancia causada por una combinación lineal de atributos. Es usada generalmente en el algoritmo <span style="color:red">Hyperellipsoidal Clustering</span></li>
    <div style="height:30px"></div>
    <li><b>Distancia Hamming</b><br/>
<img src="img/hamming.png" width="160px" style="display:block;margin:0 auto;"/><br/>
La distancia de Hamming es el número de valores que son diferentes entre dos vectores. Por lo general, se usa para comparar dos cadenas binarias de igual longitud. También se puede usar para cadenas comparando qué tan similares son entre sí calculando la cantidad de caracteres que son diferentes entre ellas.
        
<h5>Desventajas</h5>
Como era de esperar, la distancia de Hamming es difícil de usar cuando dos vectores no tienen la misma longitud.<br/>
Además, no tiene en cuenta el valor real siempre que sean diferentes o iguales. Por lo tanto, no se recomienda utilizar esta medida de distancia cuando la magnitud es una medida importante.
<h5>Casos de uso</h5>
Los casos de uso típicos incluyen la corrección/detección de errores cuando los datos se transmiten a través de redes informáticas. Se puede usar para determinar el número de bits distorsionados en una palabra binaria como una forma de estimar el error.
<p>Además, también puede usar la distancia de Hamming para medir la distancia entre variables categóricas.</p></li>

    <li><b>Haversine</b><br/>
<img src="img/haversine.png" width="150px" style="display:block;margin:0 auto;"/><br/>
        Es la distancia entre dos puntos en una esfera dadas sus longitudes y latitudes. Es muy similar a la distancia Euclidiana en que calcula la línea más corta entre dos puntos. La principal diferencia es que no es posible una línea recta ya que aquí se supone que los dos puntos están sobre una esfera.
    $$d=2rarcsin \left(\sqrt{sin^2 \left( \frac{\varphi _2 - \varphi _1}{2} \right) +cos(\varphi _1) cos(\varphi _2) sin^2 \left( \frac{\lambda _2 - \lambda _1}{2} \right)} \right) $$
    <h5>Desventajas</h5>
    Una desventaja de esta medida de distancia es que se supone que los puntos se encuentran en una esfera. En la práctica, esto raramente ocurre ya que, por ejemplo, la tierra no es perfectamente redonda, lo que podría dificultar el cálculo en ciertos casos. En cambio, sería interesante mirar hacía la distancia de <span style="color:red">Vincenty</span>, que asume un elipsoide.
    <h5>Casos de uso</h5>
    Como era de esperar, la distancia de Haversine se usa a menudo en la navegación. Por ejemplo, puede usarla para calcular la distancia entre dos países cuando vuela entre ellos. Tenga en cuenta que es muy poco adecuada si las distancias en sí mismas ya no son tan grandes pues la curvatura no tendrá un impacto tan grande.</li>
    <div style="height:30px"></div>
    <li><b>Índice de Jaccard (Intersección sobre unión)</b><br/>
<img src="img/jaccard.png" width="200px" style="display:block;margin:0 auto;"/><br/>
    Se utiliza para calcular la similitud y diversidad de conjuntos de muestras.Es el tamaño de la intersección dividido por el tamaño de la unión de los conjuntos de muestras.<br/>
    En la práctica, es el número total de entidades similares entre conjuntos dividido por el número total de entidades. Por ejemplo, si dos conjuntos tienen 1 entidad en común y hay 5 entidades diferentes en total, entonces el índice de Jaccard sería <sup>1</sup>/<sub>5</sub>=0.2. Para calcular la distancia de Jaccard simplemente restamos el índice de Jaccard de 1:
    $$D(x,y)=1-\frac{\lvert x \cap y \rvert}{\lvert x \cup y \rvert}$$
    <h5>Desventajas</h5>
    Una gran desventaja del índice de Jaccard es que esta muy influenciado por el tamaño de los datos. Los grandes conjuntos de datos pueden tener un gran impacto en el índice, ya que podrían aumentar significativamente la unión y mantener la intersección similar.
    <h5>Casos de uso</h5>
    El índice de Jaccard se usa a menudo en aplicaciones donde se usan datos binarios o binarizados. Cuando tiene un modelo de <b>aprendizaje profundo</b> (<i>deep learning</i>) que predice segmentos de una imágen, por ejemplo, un automóvil, el índice de Jaccard se puede utilizar para calcular qué tan preciso es ese segmento pronosticado con etiquetas verdaderas.<br/>
    Del mismo modo, se puede utilizar en el análisis de similitud de texto para medir la cantidad de superposición de elección de palabras que existe entre los dos documentos. Por lo tanto se puede usar para comparar conjuntos de patrones.</li>
    <div style="height:30px"></div>
    <li><b>Índice de S$\phi$rensen-Dice</b><br/>
<img src="img/sorensendice.png" width="190px" style="display:block;margin:0 auto;"/><br/>
    El índice de S$\phi$rensen-Dice es muy similar al índice de Jaccard en que mide la similitud y diversidad de conjuntos de muestras.<br/>
    El índice de S$\phi$rensen-Dice es un poco más intuitivo porque se puede ver como el porcentaje de superposición entre dos conjuntos, que es un valor entre 0 y 1:
    $$D(x,y)=\frac{2 \lvert x \cap y \rvert}{\lvert x \rvert + \lvert y \rvert}$$
    <h5>Desventajas</h5>
    Al igual que el índice de Jaccard, ambos exageran la importancia de los conjuntos con poca o ninguna verdad fundamental. Como resultado, podría dominar el puntaje promedio obtenido en múltiples conjuntos. Pondera cada elemento de forma inversamente proporcional al tamaño del conjunto relevante en lugar de tratarlos por igual.
    <h5>Casos de uso</h5>
    Los casos de uso son similares (si no los mismos) que el índice de Jaccard. Lo encontrará tipicamente usado en tareas de segmentación de imágenes o análisis de similitud de texto.</li>
