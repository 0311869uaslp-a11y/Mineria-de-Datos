```python
import pandas as pd
df=pd.read_csv('cervical-cancer_csv.csv')
df.head(10)
```

```python
df["STDs (number)"]=df["STDs (number)"].replace([0,1,2,3,4],["muy bajo","bajo","medio","alto","muy alto"])
```

```python
df.head(10)
```

```python
df.isna().sum()/len(df)*100
```

```python
df.dropna(subset=['Smokes'],axis=0,inplace=True)
```

```python
df.isna().sum()/len(df)*100
```

```python
df.dropna(thresh=0.7*len(df),axis=1,inplace=True)
df.isna().sum()/len(df)*100
```

```python
df.dtypes
```

```python
import numpy as np
numeric = df.select_dtypes(include=np.number)
numeric_columns = numeric.columns
numeric_columns
```

```python
df[numeric_columns] = df[numeric_columns].fillna(df.mean(numeric_only= True))
```

```python
df.isna().sum()/len(df)*100
```

```python
boolean_columns = df.select_dtypes(include=object).columns.tolist()
```

```python
#df["steroid"]=df["steroid"].fillna(df["steroid"].mode()[0])
for i in boolean_columns:
    df[i]=df[i].fillna(df[i].mode()[0])
```

```python
df.isna().sum()/len(df)*100
```

<h1>Datos anómalos/ atípicos/ outliers</h1>

```python
import pandas as pd
import matplotlib.pyplot as plt
hepatitis=pd.read_csv('hepatitis.csv')
hepatitis.head(10)
```

```python
plt.title("Albumin")
plt.hist(hepatitis["albumin"],edgecolor="red",linewidth=1)
plt.show()
```

```python
import numpy as np
numeric1 = hepatitis.select_dtypes(include=np.number)
numeric_columns1 = numeric1.columns

hepatitis[numeric_columns1] = hepatitis[numeric_columns1].fillna(hepatitis.mean(numeric_only= True))
hepatitis.isna().sum()/len(hepatitis)*100
```

```python
hepatitis.head(10)
```

```python
plt.boxplot(hepatitis["albumin"],vert=False)
plt.grid(True)
plt.xticks(np.arange(0,8,1))
plt.show()
```

```python
q1=hepatitis["albumin"].quantile(0.25)
print("Primer cuartil:\t\t",q1)
q3=hepatitis["albumin"].quantile(0.75)
print("Tercer cuartil:\t\t",q3)
IQR=q3-q1
print("Rango intercuartil:\t",IQR)
```

```python
BI=q1-1.5*IQR
BS=q3+1.5*IQR
print("Bigote inferior:\t",BI)
print("Bigote superior:\t",BS)
```

```python
uos=(hepatitis["albumin"]<BI) | (hepatitis["albumin"]>BS)
print("Ubicacion outliers:\n",uos)
```

```python
outliers=hepatitis[uos]
outliers
```

```python
uso=(hepatitis["albumin"]>=BI) & (hepatitis["albumin"]<=BS)
so=hepatitis[uso]
so
```

```python
hepatitis.drop(hepatitis.loc[(hepatitis["albumin"]<BI) | (hepatitis["albumin"]>BS)].index, inplace=True)
```

```python
plt.title("Albumin sin outliers")
plt.hist(hepatitis["albumin"],edgecolor="red",linewidth=1)
plt.show()
```

```python
plt.boxplot(hepatitis["albumin"],vert=False)
plt.grid(True)
plt.xticks(np.arange(0,8,1))
plt.show()
```

```python
import numpy as np
numerich = hepatitis.select_dtypes(include=np.number)
numeric_columnsh= numerich.columns
import pandas as pd
hepatitis=pd.read_csv('hepatitis.csv')
```

```python
hepatitis[numeric_columnsh] = hepatitis[numeric_columnsh].fillna(hepatitis.mean(numeric_only= True))
```

```python
hepatitis.isna().sum()/len(hepatitis)*100
```

```python
hepatitis
```

```python
columnas=["age", "bilirubin", "alk_phosphate", "sgot", "albumin", "protime"]
for i in columnas:
    print(i)
    print("Media: ",hepatitis[i].mean(),"\nDesviación estandar: ",hepatitis[i].std())#media y desviacion estandar
    plt.title(i)
    plt.hist(hepatitis[i])
    plt.show()
    #############################################################################################################
    hepatitiz=(hepatitis[i]-hepatitis[i].mean())/hepatitis[i].std()#estandarizacion z-score
    plt.title(i)
    plt.hist(hepatitiz)
    plt.show()
    #############################################################################################################
    plt.title(i)
    plt.boxplot(hepatitis[i],vert=False)
    plt.grid(True)
    plt.show()
    print("\n##########################################################################################################")
    print("\n")
```
