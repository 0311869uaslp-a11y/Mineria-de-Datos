<h1>Preprocesamiento de los Datos</h1>
<h2>Integración: covarianza y correlación</h2>
La covarianza y la correlación son muy útiles para comprender la relación entre dos variables continuas. 
<ul>
    <li>La <span style="color:red">covarianza</span> indica si ambas variables varían en la misma dirección (covarianza positiva) o en dirección opuesta (covarianza negativa).<br/>
        No hay importancia en el valor numérico de covarianza, solo el signo es útil. </li>
    <li>La <span style="color:red">correlación</span> explica sobre el cambio en una variable, indica cuánto cambio de proporción en la segunda variable.<br/>
        La correlación varía entre -1 a +1. <br/>
        Si el valor de correlación es 0, significa que no existe una relación lineal entre las variables, sin embargo, puede existir otra relación funcional.</li>
</ul>

<h3>Covarianza muestral</h3>
En el estudio de la covarianza sólo el signo importa. El valor positivo muestra que ambas variables varían en la misma dirección y el valor negativo muestra que varían en la dirección opuesta.<br/>
La covarianza entre dos variables x e y se puede calcular de la siguiente manera:

<ul>
    <li>x̄ es la media muestral de x</li>
    <li>ȳ es la media muestral de y</li>
    <li>x<sub>i</sub> e y<sub>i</sub> son los valores de x e y para el registro i-ésimo en la muestra</li>
    <li>N es el no de registros en la muestra</li>
</ul>

<h3>Correlación</h3>
<p>La correlación lineal es un método estadístico que permite cuantificar la relación lineal existente entre dos variables. Existen varios estadísticos, llamados coeficientes de correlación lineal, desarrollados con el objetivo de medir este tipo de asociación, algunos de los más empleados son <span style="color:red">Pearson, Spearman y Kendall</span>.</p>
<p>Con frecuencia, los estudios de correlación lineal preceden a análisis más complejos, como la creación de <span style="color:red">modelos de regresión</span>. Primero, se analiza si las variables están correlacionadas y, en caso de estarlo, se procede a generar modelos.</span><br/>
<p>Es importante destacar que, la existencia de correlación entre dos variables, no implica necesariamente causalidad. La asociación observada puede deberse a un tercer factor (confounder).</p>
Los coeficientes de correlación lineal comparten las siguientes características:
<ul>
    <li>Su valor está comprendido en el rango [+1 , -1]. Siendo +1 una correlación positiva perfecta y -1 una correlación negativa perfecta.</li>
    <li>Se emplean como medida de la fuerza de asociación entre dos variables (tamaño del efecto):</li>
    <ul>
        <li>0: asociación nula.</li>
        <li>0.1: asociación pequeña.</li>
        <li>0.3: asociación mediana.</li>
        <li>0.5: asociación moderada.</li>
        <li>0.7: asociación alta.</li>
        <li>0.9: asociación muy alta.</li>
    </ul>
</ul>
<img src="img\correlacion.png" style="margin:0 auto"/>
Desde el punto de vista práctico, las principales diferencias entre estos tres coeficientes son:
<ul>
    <li>La correlación de <span style="color:red">Pearson</span> funciona bien con variables cuantitativas que tienen una distribución normal o próxima a la normal. Es más sensible a los valores extremos que las otras dos alternativas.</li>
    <li>La correlación de <span style="color:red">Spearman</span> se emplea con variables cuantitativas (continuas o discretas). En lugar de utilizar directamente el valor de cada variable, los datos son ordenados y reemplazados por su respectivo orden ranking. Es un método no paramétrico muy utilizado cuando no se satisface la condición de normalidad necesaria para aplicar la correlación de Pearson.</li>
    <li>La correlación de <span style="color:red">Kendall</span> es otra alternativa no paramétrica que, al igual que la correlación de Spearman, utiliza la ordenación de las observaciones ranking. Es recomendable cuando se dispone de pocos datos y muchos de ellos ocupan la misma posición en el rango, es decir, cuando hay muchas ligaduras.</li>
</ul>
</p>

<h3>Significancia estadística</h3>
<p>Además del valor obtenido para el coeficiente de correlación, es necesario calcular su significancia estadística. Por muy cercano que sea el valor del coeficiente de correlación a  +1 o −1 si no es significativo, no se dispone de evidencias suficiente para afirmar que existe una correlación real, ya que el valor observado podría deberse a simple aleatoriedad.</p>
<p>El test paramétrico de significancia estadística empleado para el coeficiente de correlación es el t-test. Donde el estadístico t se obtiene acorde a la ecuación:
<img src="img/ttest.png" width="250px" style="margin:0 auto" /><br/>
    Donde:
    <ul>
        <li>r es el valor del coeficiente de correlación</li>
        <li>N es el número de observaciones disponibles.</li>
        <li>Los grados de libertad se calculan como  df=N−2</li>
    </ul>
</p>
<p>En este test, se considera como hipótesis nula (H<sub>0</sub>) que las variables son independientes (coeficiente de correlación poblacional = 0), y como hipótesis alternativa (H<sub>a</sub>), que sí existe relación (coeficiente de correlación poblacional≠0)</p>
<p>También se puede calcular la significancia de un coeficiente de correlación recurriendo a métodos no paramétricos como el <span style="color:red">algoritmo bootstrapping</span>.</p>

<h3>Tamaño de efecto</h3>
La correlación lineal, además del valor del coeficiente de correlación y de sus significancia, también tiene un tamaño de efecto asociado conocido como coeficiente de determinación R<sup>2</sup>.
<ul>
    <li>R<sup>2</sup> se interpreta como la cantidad de varianza de Y explicada por  X.</li>
    <li>En el caso del coeficiente de Pearson y el de Spearman, R<sup>2</sup> se obtiene elevando al cuadrado el coeficiente de correlación.<br/>
    En el caso de Kendall no se puede calcular de este modo.</li>
</ul>
<h3>Coeficiente de Pearson</h3>
El coeficiente de correlación de Pearson es la covarianza estandarizada.<br/>
<img src="img\pearson.png" width="300px" style="margin:0 auto"/><br/>
Donde:<br/>
<ul>
    <li>&sigma;x,&sigma;y son las desviaciones estándar de los conjuntos x,y respectivamente</li>
</ul>
La anterior ecuación se corresponde con el coeficiente de Pearson poblacional (&rho;). En la práctica, raramente se tiene acceso a toda la población, por lo que su valor se estima a partir de una muestra mediante el coeficiente de Pearson muestral (r).
<img src="img/pearson_muestral.png" width="500px" style="margin:0 auto"/>
<h5>Condiciones</h5>
Las condiciones que se deben de cumplir para que el coeficiente de correlación de Pearson sea válido son:
<ul>
    <li>La relación que se quiere estudiar es de tipo lineal (de lo contrario, el coeficiente de Pearson no la puede detectar).</li>
    <li>Las dos variables deben de ser numéricas.</li>
    <li>Normalidad: ambas variables se tienen que distribuir de forma normal. En la práctica, se suele considerar válido aun cuando se alejan moderadamente de la normalidad.</li>
    <li>Homocedasticidad: la varianza de Y debe ser constante a lo largo de la variable X. Esto se puede contrastar si en un scatterplot los valores de Y mantienen la misma dispersión en las distintas zonas de la variable  X.</li>
</ul>
<h5>Características</h5>
<ul>
    <li>Toma valores entre [-1, +1], siendo +1 una correlación lineal positiva perfecta y -1 una correlación lineal negativa perfecta.</li>
    <li>Es independiente de las escalas en las que se midan las variables.</li>
    <li>No varía si se aplican transformaciones a las variables.</li>
    <li>No tiene en consideración que las variables sean dependientes o independientes.</li>
    <li>El coeficiente de correlación de Pearson no equivale a la pendiente de la recta de regresión.</li>
    <li>Es sensible a <span style="color:red">outliers</span> (valores atípicos), por lo que se recomienda, en caso de poder justificarlos, excluirlos antes de realizar el cálculo.</li>
</ul>

<h3>Coeficiente de Spearman (Spearman's rho)</h3>
El coeficiente de Spearman es el equivalente al coeficiente de Pearson pero con una previa transformación de los datos a <span style="color:red">ranking</span>. Esto significa que, en lugar de utilizar directamente el valor de cada variable, los datos son ordenados y reemplazados por su respectivo orden (primero, segundo, tercero...). Se emplea como alternativa no paramétrica al coeficiente de Pearson cuando los valores son ordinales, o bien, cuando los valores son continuos pero no satisfacen la condición de <span style="color:red">normalidad</span>.
<img src="img\sperman.png" width="320px"  style="margin:0 auto"/>
<b>Donde:</b><br/>
<ul>
    <li>d<sub>i</sub> la distancia entre los rangos de cada observación (x<sub>i</sub>−y<sub>i</sub>)</li>
    <li>n el número de observaciones</li>
</ul>
<p>El coeficiente de Spearman requiere que la relación entre las variables sea <span style="color:red">monótona</span>, es decir, que cuando una variable crece la otra también lo hace o cuando una crece la otra decrece (que la tendencia sea constante).</p>
<p>Al trabajar con el orden de las observaciones en lugar de su valor real, tiene la característica de ser menos sensible que Pearson a valores extremos.</p>

<h3>Coeficiente Tau de Kendall</h3>
<p>La correlación de Kendall es un método no paramétrico que, al igual que la correlación de Spearman, utiliza la ordenación de las observaciones ranking.</p>
<p>Es otra alternativa al coeficiente de correlación de Pearson cuando no se cumple la condición de normalidad. Suele utilizarse en lugar del coeficiente de Spearman cuando el número de observaciones es pequeño o los valores se acumulan en una región por lo que el número de ligaduras al generar el ranking es alto.</p>
<img src="img\kendall.png" width="220px" style="margin:0 auto"/>
<b>Donde:</b>
<ul>
    <li>C es el número de pares concordantes, aquellos en los que el rango de la segunda variable es mayor que el rango de la primera variable</li>
    <li>D es el número de pares discordantes, cuando el rango de la segunda es igual o menor que el rango de la primera variable.</li>
</ul>

```python
# Tratamiento de datos
# ==============================================================================
import pandas as pd
import numpy as np
from sklearn.datasets import load_diabetes

# Gráficos
# ==============================================================================
import matplotlib.pyplot as plt
from matplotlib import style
import seaborn as sns

# Preprocesado y análisis
# ==============================================================================
import statsmodels.api as sm
import pingouin as pg
from scipy import stats
from scipy.stats import pearsonr

# Configuración matplotlib
# ==============================================================================
plt.style.use('ggplot')

# Configuración warnings
# ==============================================================================
import warnings
warnings.filterwarnings('ignore')
```

```python
url = ('https://raw.githubusercontent.com/JoaquinAmatRodrigo/' +
       'Estadistica-machine-learning-python/master/data/Howell1.csv')
datos = pd.read_csv(url)

# Se utilizan únicamente información de individuos mayores de 18 años.
datos = datos[datos.age > 18]

datos.info()
```

<h4>Análisis gráfico</h4>
En primer lugar se representan las dos variables mediante un diagrama de dispersión (scatterplot) para intuir si existe relación lineal o monotónica. Si no la hay, no tiene sentido calcular este tipo de correlaciones.

```python
# Gráfico
# ==============================================================================
fig, ax = plt.subplots(1, 1, figsize=(6,4))
ax.scatter(x=datos.height, y=datos.weight, alpha= 0.8)
ax.set_xlabel('Altura')
ax.set_ylabel('Peso');
```

<p>El diagrama de dispersión parece indicar una relación lineal positiva entre ambas variables.</p>
<p>Para poder elegir el coeficiente de correlación adecuado, se tiene que analizar el <span style="color:blue">tipo de variables y la distribución</span> que presentan. En este caso, ambas variables son <span style="color:red">cuantitativas continuas</span> y pueden ordenarse para convertirlas en un <span style="color:red">ranking</span>, por lo que, a priori, los tres coeficientes podrían aplicarse. La elección se hará en función de la distribución que presenten las observaciones: <span style="color:red">normalidad, homocedasticidad</span> y presencia de <span style="color:red">outliers</span>.
<h4>Normalidad</h4>

```python
# Gráfico distribución variables
# ==============================================================================
fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))

axs[0].hist(x=datos.height, bins=20, color="#3182bd", alpha=0.5)
axs[0].plot(datos.height, np.full_like(datos.height, -0.01), '|k', markeredgewidth=1)
axs[0].set_title('Distribución altura (height)')
axs[0].set_xlabel('height')
axs[0].set_ylabel('counts')

axs[1].hist(x=datos.weight, bins=20, color="#3182bd", alpha=0.5)
axs[1].plot(datos.weight, np.full_like(datos.weight, -0.01), '|k', markeredgewidth=1)
axs[1].set_title('Distribución peso (weight)')
axs[1].set_xlabel('weight')
axs[1].set_ylabel('counts')


plt.tight_layout();
```

```python
# Gráfico Q-Q
# ==============================================================================
fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))

sm.qqplot(
    datos.height,
    fit   = True,
    line  = 'q',
    alpha = 0.4,
    lw    = 2,
    ax    = axs[0]
)
axs[0].set_title('Gráfico Q-Q height', fontsize = 10, fontweight = "bold")
axs[0].tick_params(labelsize = 7)

sm.qqplot(
    datos.weight,
    fit   = True,
    line  = 'q',
    alpha = 0.4,
    lw    = 2,
    ax    = axs[1]
)
axs[1].set_title('Gráfico Q-Q weight', fontsize = 10, fontweight = "bold")
axs[1].tick_params(labelsize = 7)
```

<p>Además del estudio gráfico, se recurre a dos test estadísticos que contrasten la normalidad de los datos: <span style="color:red">Shapiro-Wilk test y D'Agostino's K-squared test</span>. Este último es el que incluye el summary de statsmodels bajo el nombre de Omnibus.</p>
<p>En ambos test, la hipótesis nula considera que los datos siguen una distribución normal, por lo tanto, si el <span style="color:red">p-value</span> no es inferior al nivel de referencia alpha seleccionado, no hay evidencias para descartar que los datos se distribuyen de forma normal.</p>

```python
# Normalidad de los residuos Shapiro-Wilk test
# ==============================================================================
shapiro_test = stats.shapiro(datos.height)
print(f"Variable height: {shapiro_test}")
shapiro_test = stats.shapiro(datos.weight)
print(f"Variable weight: {shapiro_test}")
```

```python
# Normalidad de los residuos D'Agostino's K-squared test
# ==============================================================================
k2, p_value = stats.normaltest(datos.height)
print(f"Variable height: Estadítico = {k2}, p-value = {p_value}")
k2, p_value = stats.normaltest(datos.weight)
print(f"Variable weight: Estadítico = {k2}, p-value = {p_value}")
```

El análisis gráfico y los test estadísticos muestran evidencias de que no se puede asumir normalidad en ninguna de las dos variables. Siendo estrictos, este hecho excluye la posibilidad de utilizar el coeficiente de Pearson, dejando como alternativas el de Spearman o Kendall. Sin embargo, dado que la distribución no se aleja mucho de la normalidad y de que el coeficiente de Pearson tiene cierta robustez, a fines prácticos sí que se podría utilizar siempre y cuando se tenga en cuenta este hecho y se comunique en los resultados. Otra posibilidad es tratar de transformar las variables para mejorar su distribución, por ejemplo, aplicando el logaritmo.

```python
# Transformación logarítmica de los datos
# ==============================================================================
fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(6, 4))

sm.qqplot(
    np.log(datos.height),
    fit   = True,
    line  = 'q',
    alpha = 0.4,
    lw    = 2,
    ax    = ax
)
ax.set_title('Gráfico Q-Q log(height)', fontsize = 13)
ax.tick_params(labelsize = 7)


shapiro_test = stats.shapiro(np.log(datos.height))
print(f"Variable height: {shapiro_test}")
```

```python
# Gráfico distribución variables
# ==============================================================================
fig, axs = plt.subplots(nrows=1, ncols=2, figsize=(10, 4))

axs[0].hist(x=datos.height, bins=20, color="#3182bd", alpha=0.5)
axs[0].plot(datos.height, np.full_like(datos.height, -0.01), '|k', markeredgewidth=1)
axs[0].set_title('Distribución altura (height)')
axs[0].set_xlabel('height')
axs[0].set_ylabel('counts')

axs[1].hist(x=np.log(datos.height), bins=20, color="#3182bd", alpha=0.5)
axs[1].plot(np.log(datos.height), np.full_like(np.log(datos.height), -0.01), '|k', markeredgewidth=1)
axs[1].set_title('Distribución log height')
axs[1].set_xlabel('height')
axs[1].set_ylabel('counts')


plt.tight_layout();
```

La trasformación logarítmica de la variable altura (height) consigue una distribución más próxima a la normal.

<h5>Pandas</h5>
Pandas permite calcular la correlación de dos Series (columnas de un DataFrame). El cálculo se hace por pares, eliminando automáticamente aquellos con valores <span style="color:blue">NaN/null</span>. Una limitación de Pandas es que no calcula la <span style="color:red">significancia estadística</span>.

```python
# Cálculo de correlación con Pandas
# ==============================================================================
print('Correlación Pearson: ', datos['weight'].corr(datos['height'], method='pearson'))
print('Correlación spearman: ', datos['weight'].corr(datos['height'], method='spearman'))
print('Correlación kendall: ', datos['weight'].corr(datos['height'], method='kendall'))
```

<h5>Scypy.stats</h5>
La implementación de Scypy.stats sí permite calcular la significancia estadística además del coeficiente de correlación. La función <span style="color:green">stats.pearsonr</span>(), devuelve un error si alguna de las observaciones contienen valores NA/null. Las funciones <span style="color:green">stats.spearmanr</span>() y <span style="color:green">stats.kendalltau</span>() sí permiten excluirlos de forma automática si se indica <span style="color:green">nan_policy='omit'</span>.

```python
# Cálculo de correlación y significancia con Scipy
# ==============================================================================
r, p = stats.pearsonr(datos['weight'], datos['height'])
print(f"Correlación Pearson: r={r}, p-value={p}")

r, p = stats.spearmanr(datos['weight'], datos['height'])
print(f"Correlación Spearman: r={r}, p-value={p}")

r, p = stats.kendalltau(datos['weight'], datos['height'])
print(f"Correlación Pearson: r={r}, p-value={p}")
```

<h5>Pingouin</h5>
La librería Pingouin tiene una de las implementaciones más completas. Con la función <span style="color:green">corr</span>() se obtiene, además del coeficiente de correlación, su significancia, intervalo de confianza y poder estadístico entre otros.

```python
# Cálculo de correlación, significancia e intervalos con pingouin
# ==============================================================================
display(pg.corr(datos['weight'], datos['height'], method='pearson'))
display(pg.corr(datos['weight'], datos['height'], method='spearman'))
display(pg.corr(datos['weight'], datos['height'], method='kendall'))
```

<h4>Conclusión</h4>
Los test estadísticos muestran una correlación lineal entre moderada y alta, con claras evidencias estadísticas de que la relación observada no se debe al azar (pvalue≈0).

<h3>Ejemplo matriz de correlaciones</h3>
<p>Cuando se dispone de múltiples variables numéricas, por ejemplo en problemas de modelado estadístico y machine learning, es conveniente estudiar el grado de correlación entre las variables disponibles.</p>
<p>Una forma de hacerlo es mediante matrices de correlación, en las que se muestra el coeficiente de correlación para cada par de variables.</p>

```python
# Datos
# ==============================================================================
url = ('https://raw.githubusercontent.com/JoaquinAmatRodrigo/' +
       'Estadistica-machine-learning-python/master/data/SaratogaHouses.csv')
datos = pd.read_csv(url, sep=",")

# Se renombran las columnas para que sean más descriptivas
datos.columns = ["precio", "metros_totales", "antiguedad", "precio_terreno",
                 "metros_habitables", "universitarios", "dormitorios", 
                 "chimenea", "banyos", "habitaciones", "calefaccion",
                 "consumo_calefacion", "desague", "vistas_lago",
                 "nueva_construccion", "aire_acondicionado"]
       
# Variables numéricas
datos = datos.select_dtypes(include=['float64', 'int'])
```

```python
# Matriz de correlación
# ==============================================================================
corr_matrix = datos.corr(method='pearson')
corr_matrix
```

Las matrices de correlación tienen el inconveniente de tener un tamaño notable cuando se dispone de muchas variables. Para facilitar la identificación de pares de variables con correlaciones altas, es conveniente convertirlas en formato de tabla larga (tidy).

```python
def tidy_corr_matrix(corr_mat):
    '''
    Función para convertir una matriz de correlación de pandas en formato tidy.
    '''
    corr_mat = corr_mat.stack().reset_index()
    corr_mat.columns = ['variable_1','variable_2','r']
    corr_mat = corr_mat.loc[corr_mat['variable_1'] != corr_mat['variable_2'], :]
    corr_mat['abs_r'] = np.abs(corr_mat['r'])
    corr_mat = corr_mat.sort_values('abs_r', ascending=False)
    
    return(corr_mat)

tidy_corr_matrix(corr_matrix).head(10)
```

```python
# Heatmap matriz de correlaciones
# ==============================================================================
fig, ax = plt.subplots(nrows=1, ncols=1, figsize=(5, 5))

sns.heatmap(
    corr_matrix,
    annot     = True,
    cbar      = False,
    annot_kws = {"size": 8},
    vmin      = -1,
    vmax      = 1,
    center    = 0,
    cmap      = sns.diverging_palette(20, 220, n=200),
    square    = True,
    ax        = ax
)

ax.set_xticklabels(
    ax.get_xticklabels(),
    rotation = 45,
    horizontalalignment = 'right',
)

ax.tick_params(labelsize = 10)
```

<h3>Correlación parcial</h3>
Como se ha explicado, la correlación estudia la relación (lineal o monotónica) existente entre dos variables. Puede ocurrir que, la relación que muestran dos variables, se deba a una tercera variable que influye sobre las otras dos. A este fenómeno se le conoce como <span style="color:red">confounding</span>. Por ejemplo, si se correlaciona el tamaño del pie de una persona con su inteligencia, se encuentra una correlación positiva alta. Sin embargo, dicha relación se debe a una tercera variable que está relacionada con las otras dos, la edad. La correlación parcial permite estudiar la relación lineal entre dos variables bloqueando el efecto de una tercera (o más) variables. Si el valor de correlación de dos variables es distinto al valor de correlación parcial de esas mismas dos variables, cuando se controla una tercera, significa que la tercera variable influye en las otras dos. La función en <span style="color:green">partial_corr</span>() del paquete <span style="color:purple">pingouin</span> permite estudiar correlaciones parciales.

<h3>Ejemplo correlación parcial</h3>
Se quiere estudiar la relación entre las variables <span style="color:blue">precio</span> y <span style="color:blue">peso</span> de los automóviles. Se sospecha que esta relación podría estar influenciada por la variable <span style="color:blue">potencia del motor</span>, ya que a mayor peso del vehículo se requiere mayor potencia y, a su vez, motores más potentes son más caros.

<h4>Datos</h4>
Para este ejemplo se utiliza el set de datos <span style="color:blue">Cars93</span> disponible en el paquete de R MASS.

```python
url = ('https://raw.githubusercontent.com/JoaquinAmatRodrigo/' +
       'Estadistica-machine-learning-python/master/data/Cars93.csv')
datos = pd.read_csv(url)
datos['log_Price'] = np.log(datos.Price)
```

```python
datos.head(10)
```

```python
# Gráfico
# ==============================================================================
fig, ax = plt.subplots(1, 1, figsize=(6,4))
ax.scatter(x=datos.Weight, y=datos.log_Price, alpha= 0.8)
ax.set_xlabel('Peso')
ax.set_ylabel('Log(Precio)');
```

El gráfico muestra una clara relación lineal entre el peso de un coche y el logaritmo de su precio.

<h4>Correlación</h4>

```python
# Cálculo de correlación lineal
# ==============================================================================
pg.corr(x=datos['Weight'], y=datos['log_Price'], method='pearson')
```

```python
# Cálculo de correlación lineal parcial
# ==============================================================================
pg.partial_corr(data=datos, x='Weight', y='log_Price', covar='Horsepower', method='pearson')
```

<h3>Conclusión</h3>
<p>La correlación entre el peso y el logaritmo del precio es alta (r=0.764) y significativa (pvalue≈0). Sin embargo, cuando se estudia su relación bloqueando la variable potencia de motor, a pesar de que la relación sigue siendo significativa, pasa a ser baja (r=0.4047).<p>
<p>Se puede afirmar que, la relación lineal existente entre el peso y el logaritmo del precio, está influenciada por el efecto de la variable potencia de motor.</p>
